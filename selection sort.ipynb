```python
def find_idx(a):
    l=len(a)
    min_idx=0
    for i in range (0, l):
        if a[i]<a[min_idx]:
            min_idx=i
    return min_idx
```

```python
def sel_sort(a):
    result=[]
    while a:
        min_idx=find_idx(a)
        result.append(a[min_idx])
        a.pop(min_idx)
    return result
```
