```python
import pandas as pd
import seaborn as sns

df= sns.load_dataset('titanic')
```

```python
df
```

```python
df[['sex','survived']].groupby(['sex']).mean().sort_values(by='survived',ascending=False)
```

```python
import matplotlib.pyplot as plt
```

```python
g1 =sns.FacetGrid(df, col='sex')
g1.map(plt.hist,'fare')
```

```python
g2=sns.FacetGrid(df,col='sex',row='embark_town')
g2.map(plt.hist,'fare')
```

```python
g4 =sns.FacetGrid(df, row='embark_town')
g4.map(sns.pointplot,'pclass','survived')
```

```python
g5 =sns.FacetGrid(df, row='embark_town')
g5.map(sns.pointplot,'pclass','survived','sex')
```

```python
g6 =sns.FacetGrid(df, col='survived',row='embark_town')
g6.map(sns.barplot,'sex','age',alpha=0.3)
g6.add_legend()
```
